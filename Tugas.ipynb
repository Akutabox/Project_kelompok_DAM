```python

import pandas as pd
import matplotlib.pyplot as plt

# Parameter simulasi
storage_awal = 1.96  # GB, termasuk cadangan awal 30%
pertumbuhan = 0.10   # 10% per bulan
jumlah_bulan = 12

# Menyimpan hasil perhitungan
hasil = []

for bulan in range(jumlah_bulan + 1):
    storage = storage_awal * (1 + pertumbuhan) ** bulan

    hasil.append({
        "Bulan": bulan,
        "Storage (GB)": round(storage, 2)
    })

# Membuat DataFrame
df_storage = pd.DataFrame(hasil)

# Menampilkan tabel
print("ESTIMASI STORAGE SELAMA 1 TAHUN")
print(df_storage.to_string(index=False))

# Menghitung hasil akhir
storage_akhir = storage_awal * (1 + pertumbuhan) ** jumlah_bulan

print(f"\nStorage awal  : {storage_awal:.2f} GB")
print(f"Pertumbuhan   : {pertumbuhan * 100:.0f}% per bulan")
print(f"Storage akhir : {storage_akhir:.2f} GB")

# Membuat grafik
plt.figure(figsize=(9, 5))
plt.plot(
    df_storage["Bulan"],
    df_storage["Storage (GB)"],
    marker="o"
)

plt.title("Pertumbuhan Storage Peta Kuliner Denpasar")
plt.xlabel("Bulan")
plt.ylabel("Storage (GB)")
plt.xticks(range(jumlah_bulan + 1))
plt.grid(True)
plt.tight_layout()
plt.show()

# Menyimpan hasil ke CSV
df_storage.to_csv("estimasi_storage_1_tahun.csv", index=False)

print("\nData berhasil disimpan ke estimasi_storage_1_tahun.csv")
```
