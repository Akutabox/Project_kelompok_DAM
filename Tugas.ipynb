```python

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ==========================================
# TEMA: PETA & ULASAN WARUNG KULINER DENPASAR
# ==========================================

# 1. DAFTAR JENIS DATA
jenis_data = pd.DataFrame({
    "No": [1, 2, 3, 4, 5, 6, 7],
    "Jenis Data": [
        "Data warung", "Data lokasi", "Data menu dan harga",
        "Data foto", "Data ulasan", "Data pengguna",
        "Data kategori"
    ],
    "Contoh Data": [
        "Nama, alamat, telepon, jam buka",
        "Latitude dan longitude",
        "Nama menu, kategori, harga",
        "Foto warung dan makanan",
        "Rating, komentar, tanggal",
        "ID dan nama tampilan",
        "Jenis makanan atau kuliner"
    ]
})

# 2. KARAKTERISTIK DATA MENGGUNAKAN 5W+1H
karakteristik = pd.DataFrame([
    ["Data warung", "Informasi warung", "Pemilik/admin",
     "Database", "Saat daftar atau berubah",
     "Memberi informasi warung", "Formulir"],
    ["Data lokasi", "Koordinat lokasi", "Pemilik/admin",
     "Peta digital", "Saat lokasi diperbarui",
     "Memudahkan pencarian", "GPS atau input koordinat"],
    ["Menu dan harga", "Nama menu dan harga", "Pemilik",
     "Database menu", "Saat menu berubah",
     "Memberi informasi harga", "Formulir menu"],
    ["Foto", "Foto warung dan makanan", "Pemilik/admin",
     "Penyimpanan file", "Saat foto diunggah",
     "Memberi gambaran visual", "Unggah gambar"],
    ["Ulasan", "Rating dan komentar", "Pengunjung",
     "Database ulasan", "Setelah kunjungan",
     "Membantu menilai warung", "Formulir ulasan"],
    ["Pengguna", "ID dan nama tampilan", "Pengunjung",
     "Database pengguna", "Saat registrasi",
     "Mengelola akun", "Formulir registrasi"],
    ["Kategori", "Jenis kuliner", "Admin",
     "Database kategori", "Saat kategori berubah",
     "Memudahkan penyaringan", "Daftar kategori"]
], columns=[
    "Jenis Data", "What (Apa?)", "Who (Siapa?)",
    "Where (Di mana?)", "When (Kapan?)",
    "Why (Mengapa?)", "How (Bagaimana?)"
])

# 3. ASUMSI JUMLAH UNIT DAN UKURAN RATA-RATA
# Semua ukuran diubah ke KB.
# Asumsi 1 MB = 1000 KB dan 1 GB = 1000 MB.

asumsi = pd.DataFrame({
    "Jenis Data": [
        "Data warung", "Data lokasi", "Menu dan harga",
        "Foto", "Ulasan", "Pengguna", "Kategori"
    ],
    "Jumlah Unit": [
        200, 200, 2000, 1000, 2000, 1500, 20
    ],
    "Ukuran Rata-rata": [
        "5 KB", "1 KB", "2 KB",
        "1.5 MB", "2 KB", "1 KB", "1 KB"
    ],
    "Ukuran per Unit (KB)": [
        5, 1, 2, 1500, 2, 1, 1
    ]
})

asumsi["Total (MB)"] = (
    asumsi["Jumlah Unit"]
    * asumsi["Ukuran per Unit (KB)"] / 1000
)

# 4. TOTAL STORAGE DAN CADANGAN 30%
total_data_mb = asumsi["Total (MB)"].sum()
cadangan_mb = total_data_mb * 0.30
storage_awal_mb = total_data_mb + cadangan_mb

ringkasan = pd.DataFrame({
    "Komponen": [
        "Total data awal",
        "Cadangan 30%",
        "Total storage termasuk cadangan"
    ],
    "Storage (MB)": [
        total_data_mb,
        cadangan_mb,
        storage_awal_mb
    ],
    "Storage (GB)": [
        total_data_mb / 1000,
        cadangan_mb / 1000,
        storage_awal_mb / 1000
    ]
})

# 5. PERTUMBUHAN STORAGE SELAMA 12 BULAN
pertumbuhan = 0.10
storage_awal_gb = storage_awal_mb / 1000
hasil = []

for bulan in range(13):
    storage = storage_awal_gb * (1 + pertumbuhan) ** bulan

    hasil.append({
        "Bulan": bulan,
        "Storage (GB)": round(storage, 3)
    })

df_pertumbuhan = pd.DataFrame(hasil)

# ==========================================
# MENAMPILKAN SEMUA TABEL
# ==========================================

print("\n1. DAFTAR JENIS DATA")
display(jenis_data)
print(jenis_data.to_string(index=False))

print("\n2. KARAKTERISTIK DATA (5W+1H)")
display(karakteristik)
print(karakteristik.to_string(index=False))

print("\n3. ASUMSI JUMLAH UNIT DAN UKURAN")
display(asumsi)
print(asumsi.to_string(index=False))

print("\n4. ESTIMASI STORAGE DAN CADANGAN 30%")
ringkasan_tampil = ringkasan.round(3)
display(ringkasan_tampil)
print(ringkasan_tampil.to_string(index=False))

print("\n5. PERTUMBUHAN STORAGE SELAMA 1 TAHUN")
display(df_pertumbuhan)
print(df_pertumbuhan.to_string(index=False))

print("\nRINGKASAN AKHIR")
print(f"Total data awal : {total_data_mb:.2f} MB")
print(f"Cadangan 30%    : {cadangan_mb:.2f} MB")
print(f"Storage awal    : {storage_awal_gb:.3f} GB")
print(f"Storage bulan 12: {df_pertumbuhan.iloc[12]['Storage (GB)']:.3f} GB")

# ==========================================
# GRAFIK PERTUMBUHAN STORAGE
# ==========================================

plt.figure(figsize=(9, 5))
plt.plot(
    df_pertumbuhan["Bulan"],
    df_pertumbuhan["Storage (GB)"],
    marker="o"
)

plt.title("Pertumbuhan Storage Warung Kuliner Denpasar")
plt.xlabel("Bulan")
plt.ylabel("Storage (GB)")
plt.xticks(range(13))
plt.grid(True)
plt.tight_layout()
plt.show()

# ==========================================
# MENYIMPAN HASIL KE FILE CSV
# ==========================================

jenis_data.to_csv("daftar_jenis_data.csv", index=False)
karakteristik.to_csv("karakteristik_5W1H.csv", index=False)
asumsi.to_csv("asumsi_storage.csv", index=False)
ringkasan.to_csv("ringkasan_storage.csv", index=False)
df_pertumbuhan.to_csv("pertumbuhan_storage.csv", index=False)

print("\nSemua tabel berhasil disimpan ke file CSV.")
```
